## **LOGISTIC REGRESSION - BCE - SIMPLE IMPLEMENTATION** ##

```python
import numpy as np
import math

iris = np.genfromtxt('iris_1D.csv', dtype=None, delimiter=',',skip_header=1)

x_data = iris[:,0]
y_data = iris[:,1]

print(x_data)
print(y_data)
```

Hai dòng cuối cùng trong đoạn code có chức năng trách (tách) dữ liệu từ mảng 2D thành hai biến/mảng 1D riêng biệt dựa trên hai cột đầu tiên của file CSV.

**x_data = iris[:, 0]:**

- iris: Mảng dữ liệu 2D vừa đọc từ file iris_1D.csv.

- [:, 0]: Sử dụng cú pháp slicing (cắt mảng) của NumPy:

    + Dấu : trước dấu phẩy có nghĩa là lấy tất cả các hàng (rows).
    
    + Số 0 sau dấu phẩy nghĩa là chỉ lấy cột đầu tiên (index 0).

$\rightarrow$ Biến x_data sẽ chứa toàn bộ giá trị thuộc cột thứ 1 của dataset.

**y_data = iris[:, 1]:**

Tương tự cú pháp trên:

- Dấu : nghĩa là lấy tất cả các hàng.

- Số 1 nghĩa là chỉ lấy cột thứ hai (index 1).

$\rightarrow$ Biến y_data sẽ chứa toàn bộ giá trị thuộc cột thứ 2 của dataset.

Tóm lại: Đoạn code này dùng để phân tách cột thứ nhất làm dữ liệu đầu vào ($X$) và cột thứ hai làm dữ liệu nhãn/đầu ra ($Y$) để chuẩn bị cho các bài toán xử lý dữ liệu hoặc mô hình học máy.

```python
# Define the sigmoid activation function.
# It converts any value into a probability between 0 and 1.
def sigmoid_function(z):
    return 1/ (1+math.exp(-z))

# Define Binary Cross-Entropy loss
# It measures the difference between the predicted probability y_hat and the actual label y.
    #if y=1, the formula becomes -log(y_hat)
    #if y=0, the formula becomes log(1-y_hat)

#y_hat should be a probability between 0 and 1, usually produced by the sigmoid function. Also, it cannot be exactly 0 or 1 because math.log(0) is undefined.

def loss_function(y_hat, y):
    return -y*math.log(y_hat) - (1-y)*math.log(1-y_hat)

# Calculate a prediction using the current weight and bias.
# z is the linear model output, and y_hat is the predicted probability.
def predict(x,w,b):
    z = w*x + b
    print('z: ', z)
    y_hat = sigmoid_function(z)
    return y_hat

# Calculate the gradients for the bias and weight.
# The gradients show how the parameters should change
# to reduce the loss.
def compute_gradient(x, y_hat, y):
    db = (y_hat - y)
    dw = x * (y_hat - y)
    return db, dw

# Update the weight and bias using gradient descent.
# The learning rate controls the size of each update.
def update(b,w,lr,db,dw):
    b = b - lr*db
    w = w - lr*dw
    return b, w

###The functions are only being defined here. Python does not calculate the expression yet. Later we can call it with actual values
```

```python
##some parameters

lr = 0.01
num_epochs = 1

w = -0.1
b = 0.1

for _ in range(num_epochs):
    for i in range(6):
        '''
        for i in range(6) means 0,1,2,3,4,5
        then each iteration selects one row 
        x = x_data[i]  # x_data[0], x_data[1], ..., x_data[5]
        y = y_data[i]  # y_data[0], y_data[1], ..., y_data[5]

        However the loop contains break:
        The break stops the inner loop after the first iteration, when i = 0. Therefore, only the first data point is printed.

        Remove break to process all six data points

        We are currently testing only the first data point, where i = 0.
        Since num_epochs = 1, the first data point is processed only once. If num_epochs were 3, the same first data point would be processed three times, with updated w and b each time
        '''


        #get data
        x = x_data[i]
        y = y_data[i]
        print('x, y: ', x, y)

        #predict
        y_hat = predict(x, w, b)
        print('y_hat: ', y_hat)

        #compute loss
        loss = loss_function(y_hat, y)
        print('loss: ', loss)

        #compute gradient
        db, dw = compute_gradient(x, y_hat, y)
        print('db, dw: ', db, dw)

        #update
        b,w = update(b,w,lr,db,dw)
        print('b, w: ', b,w)
        print('----------')
        break
```
